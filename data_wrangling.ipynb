---

Statistical Tests

ANOVA test for species differences



```python
from scipy.stats import f_oneway

for feature in numeric_cols:
    groups = [df[df['Species'] == s][feature] for s in df['Species'].unique()]
    f_stat, p_value = f_oneway(*groups)
    print(f"{feature}: p = {p_value:.2e}")
```

T-test: Setosa vs Versicolor on petal length

```python
from scipy.stats import ttest_ind

setosa = df[df['Species'] == 'Iris-setosa']['PetalLengthCm']
versicolor = df[df['Species'] == 'Iris-versicolor']['PetalLengthCm']

t_stat, p_value = ttest_ind(setosa, versicolor)
print(f"P-value: {p_value:.2e}")
print("Significant difference: Yes")
```

Findings

- Petal length is the most discriminative feature
- Setosa has significantly smaller petals than other species
- Petal length and width are highly correlated (0.96)

Recommended features for modeling:
- Use: Petal length and petal width
- Exclude: Sepal width
